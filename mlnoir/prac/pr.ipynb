```python
import pandas as pd
import numpy as np
```

```python
class linearreg:
    def fit(self,X,y):
        X = np.asarray(X,dtype=float)
        y = np.asarray(y,dtype=float)
        X_b = np.c_[X,np.ones(X.shape[0])]
        w = np.linalg.pinv(X_b.T @ X_b) @ X_b.T @ y
        self.weights = w[:-1]
        self.bias = w[-1]
    def predict(self,X):
        X = np.asarray(X,dtype=float)
        return X @ self.weights + self.bias
class linearreg1:
    def __init__(self,lr=0.1,epoch=1000):
        self.lr = lr
        self.epoch = epoch
    def fit(self,X,y):
        X = np.asarray(X,dtype=float)
        y = np.asarray(y,dtype=float)
        samples,features = X.shape
        self.weights = np.zeros(features)
        self.bias = 0.0

        for _ in range(self.epoch):
            ypred = X @ self.weights + self.bias
            dw = (2/samples) * (X.T @ ( ypred - y))
            db = (2/samples) * np.sum(ypred -y)
            self.weights -= self.lr * dw
            self.bias -= self.lr * db

            
class logreg:
    def __init__(self,lr=0.1,epochs=1000):
        self.lr = lr
        self.epochs = epochs
    def sigmoid(self,z):
        return 1.0 / (1.0 + np.exp(-np.clip(z,-500,500)))
    def fit(self,X,y):
        X = np.asarray(X,dtype=float)
        y = np.asarray(y,dtype=float)
        samples, features = X.shape
        self.weights = np.zeros(features)
        self.bias = 0.0

        for _ in range(self.epochs):

            z = X @ self.weights + self.bias
            ypred = self.sigmoid(z)
            dw = (1/samples) * (X.T @ (ypred - y))
            db = (1/samples) * np.sum(ypred - y)

            self.weights -= self.lr * dw
            self.bias -= self.lr * db
    def predictprob(self,X):
        X = np.asarray(X,dtype=float)
        return self.sigmoid(X @ self.weights + self.bias)
    def predict(self,X,t=0.5):
        return (self.predictprob(X) >= t).astype(int)
```
