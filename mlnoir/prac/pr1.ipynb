```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("udemy_courses_with_student_ids_and_ratings.csv")
R = pd.pivot(df,index='student_id',columns='course_id',values='course_rating')
R.head()

mean = R.mean(axis=1)
R1 = R.sub(mean,axis=0).fillna(0)
R1.head()
```

```python
U , S , Vt = np.linalg.svd(R1,full_matrices=False)
k = 2
Uk = U[:,:k]
Sk = np.diag(S[:k])
Vtk = Vt[:k,:]

R2 = Uk @ Sk @ Vtk
R2new = pd.DataFrame(R2,index=R.index,columns=R.columns).add(mean,axis=0)
R2new.head()
```
